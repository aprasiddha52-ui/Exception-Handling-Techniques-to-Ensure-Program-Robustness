```python
# FIX - keep asking until the room number is unique
rooms = [101, 102]
typed = iter([102, 101, 103])           # simulated typing: duplicate, duplicate, valid
 
number = next(typed)
while number in rooms:                  # repeats until the number is unique
    print("Room", number, "already exists.")
    number = next(typed)
print("Accepted room number:", number)
print("Is it a duplicate?", number in rooms)   # False - correct
```

```python
# FIX - date and time in the backup file name
import glob
import os
import time
from datetime import datetime
 
for text in ["Backup of Monday", "Backup of Tuesday"]:
    stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    with open(f"fix6_backup_{stamp}.txt", "w") as file:
        file.write(text)
    time.sleep(1.1)                     # so the two time stamps are different
 
files = sorted(glob.glob("fix6_backup_*.txt"))
print("Backup files:", files)           # both backups are kept
for name in files:
    os.remove(name)
```

```python
# Exception 1 - SyntaxError
import ast
 
text = "15 0"                              # the user types an invalid rate
try:
    value = ast.literal_eval(text)
except SyntaxError as error:
    print("SyntaxError handled:", error)
    print("Please enter the rate as 150 or 149.95.")
```

```python
# Exception 2 - ValueError
text = "abc"                               # the user types text instead of a number
try:
    value = int(text)
except ValueError as error:
    print("ValueError handled:", error)
    print("Please enter a whole number.")
```

```python
# Exception 3 - ZeroDivisionError
def calculate_occupancy(allocated, total):
    try:
        return allocated / total * 100
    except ZeroDivisionError as error:
        print("ZeroDivisionError handled:", error)
        print("There are no rooms, so occupancy is 0.")
        return 0.0

print("Occupancy:", calculate_occupancy(0, 0), "%")
```
